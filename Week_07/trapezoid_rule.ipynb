An integral is the area under a curve, or equivalently the accumulation of a rate: distance from velocity, velocity from acceleration, resisting force from strength along a plane.
One of the most common ways to calculate integrals numerically is using the **trapezoid rule**.

## Equation for trapezoid area

For some function, $f(x)$, consider a trapezoid whose width goes from $x=a$ to $x=b$ and whose height is $f(a)$ at $x=a$ and  $f(b)$ at $x=b$. The area of that trapezoid the width times the average height:

$$\int_a^b f(x)\,dx \approx (b - a)\,\frac{f(a) + f(b)}{2}$$
