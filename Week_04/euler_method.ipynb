A **differential equation** relates a quantity to its own rate of change. Many differential equations have no closed-form solution.

**Euler's method** is an algorithm for approximating the solution to differential equations.

## The idea

If we know the value at one point and the derivative there, the derivative is the slope of the tangent line, and a short step along the tangent is a good estimate of the next value:

$$t_{n+1} = t_n + \Delta t, \qquad y_{n+1} = y_n + \Delta t \cdot f(t_n, y_n)$$

where $f(t, y) = dy/dt$. 

Repeat from the new point and the steps trace out an approximate solution. 
We need three things: 

1. a function for the derivative, 
2. an initial condition, and 
3. a step size (which we get to choose).

## Example problem

Use the differential equation and initial condition below to find $y$ when $t=2.5$:

$$y(t) = 1 - \frac{dy}{dt},\quad y(0) = 3$$

First, rearrange the differential equation so we have a "function for the derivative":

$$\frac{dy}{dt} = 1 - y(t)$$

_Note: this differential equation has a closed-form solution ($y = 1 + 2e^{-t}$), so we can see how good the Euler approximation is._

## Another example

A tank with cross-section $A$ drains through a hole of area $a$ in its bottom while a pipe fills it at a constant rate $Q$. 

<img src="tank_sketch.png" alt="A tank filled to depth h, with inflow Q at the top and a hole of area a in the bottom" width="40%">


Torricelli's law says the water leaves at speed $\sqrt{2gh}$, so the depth $h$ obeys

$$\frac{dh}{dt} = \frac{Q}{A} - \frac{a}{A}\sqrt{2gh}$$

- $h(0) = 2$ m (_the initial condition_)
