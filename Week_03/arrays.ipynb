## Checklist for arrays

* `import numpy as np`
* `np.array(list)`, `np.linspace(start, stop, n)`, `np.arange(start, stop, step)`, `np.zeros(n)`
* arithmetic and `np.` functions apply to every element; no loop needed
* index and slice like a list: `x[0]`, `x[-1]`, `x[a:b]`
* `len(x)` or `x.shape` for the size
