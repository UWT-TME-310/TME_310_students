## Problem 1

Pile driving is the process of installing deep foundations by driving them into the ground, often with a diesel impact hammer.
The diesel hammer is essentially a single piston combustion engine.
The fuel ignition launches the hammer ram up into the air.
When the ram falls back down, it compresses and heats an air/fuel mixture in the piston, reaching its ignition temperature and starting the process over again.

[Watch an example of a diesel hammer driving a pile](https://uwt-tme-310.github.io/TME_310_A/videos/pile_driving.html) (short video, opens on the course website).

A **pile installation log** includes two key parameters that are tracked with the depth (typically in 1-foot increments):

1. the hammer blows per minute (**BPM**) and
2. the hammer blows per foot (**BPF**).

The **blows per minute** (averaged over each foot the pile is driven) provides a way to determine the height the hammer falls in free fall onto the top of the pile for each blow.
By converting BPM to time between blows ($t_{bb}$) in seconds, we can estimate the hammer fall height ($\Delta H$) using a basic free-fall kinematics equation:

$\displaystyle \Delta H = v_0 t + \frac{a}{2} \left( t \right)^2$

Since the hammer ram takes the same time to go up as it does to come down, and the velocity at the maximum height is zero, we can find the stroke height using:

$\displaystyle \Delta H = \frac{a}{2} \left( \frac{t_{bb}}{2} \right)^2$

Combined with the ram weight, the fall height gives us the energy associated with a single blow (energy = force $\times$ distance).

The **blows per foot** on a pile installation log describes the number of blows it took to drive the pile 1 foot into the ground.
Multiplying the energy per blow by BPF yields the energy used to drive the pile 1 foot.

The stability model is deliberately simple: a **planar wedge** of soil sliding on a flat failure plane that passes through the toe of the slope.
The ground is horizontal above and below the slope, and the slope face has a single angle $\beta$.

<img src="wedge_geometry.png" width="60%">

The failure plane makes an angle $\theta$ with the horizontal, where $0 < \theta < \beta$.
This is the block-on-an-inclined-plane problem from physics with two additions: the block's shape depends on $\theta$, and the block is held in place by the shear strength of the soil along the plane instead of a friction coefficient.

For a slope of height $H$ and unit weight $\gamma$ (analyzed per unit width into the page):

* Wedge area: $A = \dfrac{H^2}{2}\left(\cot\theta - \cot\beta\right)$
* Wedge weight: $W = \gamma A$
* Failure plane length: $L = \dfrac{H}{\sin\theta}$
* Driving force (component of $W$ along the plane): $D = W \sin\theta$
* Resisting force for a uniform Mohr-Coulomb soil with cohesion $c$ and friction angle $\phi$: $R = cL + W\cos\theta\tan\phi$

The **factor of safety** is $FS = R / D$. The slope is predicted to fail when $FS < 1$.
