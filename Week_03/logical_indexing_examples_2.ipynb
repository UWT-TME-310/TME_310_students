## The fraction inside estimates $\pi$

The square has area $2 \times 2 = 4$ and the circle has area $\pi \cdot 1^2 = \pi$. If darts land uniformly, the fraction inside is about $\pi / 4$, so

$$\pi \approx 4 \cdot \frac{\text{darts inside}}{\text{darts thrown}}$$

## Checklist for logical indexing

* a comparison on an array gives a boolean array (a mask)
* `x[mask]` keeps the `True` entries; `mask.sum()` counts them
* combine conditions with `&`, `|` and `~`, each comparison in parentheses
* for a question about one array, build the mask on it and apply the mask to the related array
* `x[mask] = value` changes only the selected entries
