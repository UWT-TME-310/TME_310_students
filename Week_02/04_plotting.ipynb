# Plotting

This notebook introduces plotting in Python with the Matplotlib library.

## Python libraries

Python has several built-in and external **libraries** that extend its capabilities.
The first one we'll use in this course is **Matplotlib**, a library for plotting.

### Importing libraries

Import libraries with the `import` statement.
It's common to use *aliases* as shorthand for imported modules by using the `as` keyword:

```python
import matplotlib as mpl
```

This imports the entire `matplotlib` library into your code. You can use `mpl` to refer to it.

### Partial imports

Sometimes libraries are large and we only need a small part of them.
In these cases, we can import a specific function using the `from` keyword:

```python
from math import exp   # <---- just imports the function math.exp()
```

Subcomponents of libraries are accessed with `.` syntax, so we can also import a submodule.
This is the typical way to start using Matplotlib:

```python
import matplotlib.pyplot as plt
```

The [Matplotlib documentation](https://matplotlib.org/stable/users/index.html) covers far more than we need.
Today we'll make a figure, draw lines on it with `plot()`, and label what we draw.

## Figure and axes


```python
fig, ax = plt.subplots()
```

* `fig` is the **figure**: the whole image
* `ax` is the **axes**: the region where data gets drawn

Almost everything we do to a plot is done through `ax`.

## A line plot

`ax.plot()` takes a list of x-values and a list of y-values and draws a line through the points.

```python
x = [1,2,3,4,5]
y = [3,7,5,3,6]
plt.plot(x,y)
```

```python
x = [1,2,3,4,5]
y = [3,7,5,3,6]
fig, ax = plt.subplots()
ax.plot(x,y)
```

Typing every value by hand gets old quickly. A loop can build the lists for us:

```python
x = list(range(100))
y = []
for x_value in x:
    y_value = x_value**2
    y.append(y_value)
fig, ax = plt.subplots()
ax.plot(x,y)
```

## Choosing the x-values

`range()` only counts in whole numbers. To get other spacings, scale the loop counter.

```python
x = []
y = []
for i in range(6):
    x_value = i/10
    y_value = x_value**2
    x.append(x_value)
    y.append(y_value)
fig, ax = plt.subplots()
ax.plot(x,y)
```

`plot()` draws straight lines between the points, so 6 points make a jagged curve.

More points make a smoother curve. 100 to 200 points is usually plenty.

```python
x = []
y = []
for i in range(200):
    x_value = i
    y_value = x_value**2
    x.append(x_value)
    y.append(y_value)
fig, ax = plt.subplots()
ax.plot(x,y)
```

## Labels


```python
x = []
y = []
for i in range(200):
    x_value = i
    y_value = x_value**2
    x.append(x_value)
    y.append(y_value)
fig, ax = plt.subplots()
ax.plot(x,y)
ax.set_xlabel("Time (s)")
ax.set_ylabel("Distance (m)")
ax.set_title("Distance traveled")
```

Every axis label should say **what** is plotted and in what **units**.

## More than one line

Call `ax.plot()` more than once and the lines share the same axes.
A `label` on each line and a call to `ax.legend()` tells them apart.

```python
x = []
y = []
y_linear = []
for i in range(10):
    x_value = i
    y_value = x_value**2
    y_lin = x_value
    x.append(x_value)
    y.append(y_value)
    y_linear.append(y_lin)
fig, ax = plt.subplots()
ax.plot(x,y, label="$x^2$")
ax.plot(x,y_linear, label = "$x$")
ax.legend()
```

## Reference lines

`ax.axhline()` and `ax.axvline()` draw horizontal and vertical lines across the whole plot.
They're useful for marking a limit or a target value.

```python
x = []
y = []
y_linear = []
for i in range(10):
    x_value = i
    y_value = x_value**2
    y_lin = x_value
    x.append(x_value)
    y.append(y_value)
    y_linear.append(y_lin)
fig, ax = plt.subplots()
ax.axhline(20, color="gray", linestyle="--")
ax.plot(x,y, label="$x^2$")
ax.plot(x,y_linear, label = "$x$")
ax.legend()
```
