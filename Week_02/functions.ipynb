## Starting point

The code from the plotting notebook:

```python
from math import exp
import matplotlib.pyplot as plt
```

```python
g = 9.81   # m/s^2
m = 80     # kg
c = 12     # kg/s

t = []
v = []
for i in range(201):
    t_value = 0.2 * i        # 0 to 40 s
    t.append(t_value)
    v.append(g * m / c * (1 - exp(-c * t_value / m)))

fig, ax = plt.subplots()
ax.plot(t, v)
ax.set_xlabel("time (s)")
ax.set_ylabel("velocity (m/s)")
```

To plot a second parachutist we would copy all of this and change `m`. Functions are the way out.

## Checklist for a function

* `def name(required, optional=default):` and a colon
* an indented body
* `return` the result (or results, separated by commas)
* define before you call
* call with positional arguments in order, or with keywords in any order
