## Problem

In hw02 you found the steepest incline on which a rigid block will not slide. Now hold the incline fixed and shake it. An earthquake moves the plane back and forth horizontally. Engineers describe the shaking by a **horizontal acceleration coefficient** $k$, the ground acceleration as a fraction of $g$, and treat it as a horizontal force $kW$ on the block pointing away from the slope, where $W = mg$ is the block's weight.

<img src="sliding_block_yield.png" width="55%">

That force adds to the force driving the block down the plane and reduces the normal force that friction depends on:

$$N = W\,(\cos\theta - k\sin\theta), \qquad D = W\,(\sin\theta + k\cos\theta)$$

The **factor of safety** against sliding is the available friction divided by the friction needed for equilibrium, $FS = \mu N / D$. The **yield coefficient** $k_y$ is the value of $k$ at which $FS = 1$: below it the block rides along with the plane, above it the block slides. The yield acceleration is $a_y = k_y\,g$.
