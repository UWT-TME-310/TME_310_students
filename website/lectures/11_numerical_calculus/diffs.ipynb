```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,2*np.pi)
y = np.sin(x)

dydx_np = np.gradient(y,x)
dydx_sp = sp.differentiate.derivative(np.sin, x)
```

```python
dydx_sp.df
```

```python
# plt.plot(x,dydx_np)
plt.plot(x, dydx_sp.df)
plt.show()
```
