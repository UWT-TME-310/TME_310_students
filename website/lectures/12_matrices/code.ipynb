```python
# Matrix multiplication example: (4x2) @ (2x3) = (4x3)
A = np.array([[1, 2], 
              [3, 4],
              [5, 6],
              [7, 8]])
B = np.array([[2, 1, 0], 
              [3, 2, 1]])
print(f"A has shape {A.shape}")
print(f"B has shape {B.shape}")
C = A @ B
print(f"C has shape {C.shape}")
print(C)
```

System of 3 masses and interconnected springs

```python
ka = 100 # N/mm
kb = 100 # N/mm
kc = 100 # N/mm
kd = 100 # N/mm

m1 = 200 # kg
m2 = 100 # kg
m3 = 100 # kg

g = 9.81 # m/s/s

a_r1 = [ka+kb, -kb, 0]
a_r2 = [-kb, kb+kc+kd, -kd]
a_r3 = [0, -kd, kd]

A = np.array([a_r1, a_r2, a_r3])

b = [m1*g, m2*g, m3*g]

x = linalg.solve(A, b) # mm
print(x)
```

```python
k = 10 # N/m
m1, m2, m3 = 2, 3, 2.5 # kg
g = 9.81 # m/s/s

A = k * np.array([
    [3, -2, 0],
    [-2, 3, -1],
    [0, -1, 1]])
print(A)
b = [m1*g, m2*g, m3*g]
print(b)
x = linalg.solve(A, b)
print(x)
```
